```python
import wget
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName('Q2').getOrCreate()
sc = spark.sparkContext

wget.download("https://raw.githubusercontent.com/nongaussian/class-2024-datamining/main/quiz/int-pair-input.txt", "intpair.txt")
```

```python
first_elems = sc.textFile("intpair.txt") \
    .filter(lambda line: line.strip() != "") \
    .map(lambda line: int(line.strip().strip("()").split(",")[0]))

first_pair = first_elems.first()
max_val = first_elems.reduce(lambda a, b: max(a, b))
min_val = first_elems.reduce(lambda a, b: min(a, b))
count_one = first_elems.filter(lambda x: x == 1).count()
count_two = first_elems.filter(lambda x: x == 2).count()

print("첫 번째 원소:", first_pair, "/ 최댓값:", max_val, "/ 최솟값:", min_val,
      "/ 1의 개수:", count_one, "/ 2의 개수:", count_two)
print("결과:", first_pair + max_val + min_val + count_one + count_two)
```
