# 팀과제 1 - Q1 (팀 4)

각 단어의 앞 두 글자(2-head word) 빈도를 세고, 빈도가 32 미만인 2-head word들의 빈도 합을 출력한다.

- 대소문자 구분 없음 → 소문자로 통일
- 조건에 따라 **길이가 2보다 큰 단어만** 사용 (PDF 예시에는 `IN`, `my` 같은 두 글자 단어가 있으나, 문제 조건을 우선함)
- 공백 기준 분리, 문장부호는 제거하지 않음

```python
import wget
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName('Q1').getOrCreate()
sc = spark.sparkContext

wget.download("https://raw.githubusercontent.com/nongaussian/class-2024-datamining/main/quiz/sherlock.txt", "sherlock.txt")
```

```python
total_sum = sc.textFile("sherlock.txt") \
    .flatMap(lambda line: line.split()) \
    .filter(lambda word: len(word) > 2) \
    .map(lambda word: (word.lower()[:2], 1)) \
    .reduceByKey(lambda a, b: a + b) \
    .filter(lambda pair: pair[1] < 32) \
    .map(lambda pair: pair[1]) \
    .reduce(lambda a, b: a + b)

print("결과:", total_sum)
```
